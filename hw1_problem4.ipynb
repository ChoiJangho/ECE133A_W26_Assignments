```python
import numpy as np
# Fix random seed for reproducibility
np.random.seed(0)
```

The below functions create `n` random floating points with their precision either specified as `single` or `double`. Each floating points follow a random distribution of x_i = m + n_i, where m = `target_mean`, and n_i is sampled from a uniform distribution from range [-sample_range, sample_range].

```python
def get_uniform_random_data(n, target_mean, sample_range, precision: str):
    assert precision in ['single', 'double'], "Precision must be 'single' or 'double'"
    np_dtype = np.float32 if precision == 'single' else np.float64
    # uniform distribution
    return np.array(target_mean + np.random.uniform(-sample_range, sample_range, size=n), dtype=np_dtype)
```

```python
def get_variance_unsafe(data):
    return np.mean(data**2) - np.mean(data)**2
```

```python
def get_variance_analytical(mean, sample_range):
    return (sample_range**2) / 3
```

```python
# must be large enough to get emirical variance close to analytical variance
n = int(1e6) # Number of samples
# additive noise uniformly sampled from [-sample_range, sample_range].
sample_range = 1
```

### Case 1: Dataset saved as single precision floating points

```python
target_mean_values = [1e1, 1e2, 1e3, 1e4, 1e5]
for target_mean in target_mean_values:
    data = get_uniform_random_data(n, target_mean, sample_range, precision='single')
    var_unsafe = get_variance_unsafe(data)
    var_analytical = get_variance_analytical(target_mean, sample_range)
    print(f"Single Precision Data with target_mean: {target_mean}")
    print(f"Variance Computed: {var_unsafe}")
    print(f"Analytical Variance: {var_analytical}\n")
```
